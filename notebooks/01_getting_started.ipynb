## Tutorial Goals

## About SpinLab

SpinLab is a Python package for EPR and magnetic resonance data processing.

It provides tools for:

- Data loading
- Signal processing
- Visualization
- Spectral analysis
- Workflow automation

## Loading Data

In the next section we will load an example dataset. The dataset is a echo detected field sweep dataset.

The example dataset is included in the repository under the `data/` directory. The filename is `17478-EDFS.d01`

```python
# Load example data here, X-band 3pESEEM example experiment

# import required packages
import spinlab as sl
from matplotlib.pylab import plt
plt.rcParams['font.family'] = 'DejaVu Sans'

# Define path to data
path = '../data/17478-EDFS.d01'

data = sl.load(path)

print(data)
```

## Inspecting the Dataset

Once the data is loaded, we can inspect:

- Dimensions
- Axes
- Metadata
- Signal values

```python
# Inspect dataset here

values = data.values

dims = data.dims

coords = data.coords

field = data.coords['Field']

print('-'*50)
print('Values:')
print(values)

print('-'*50)
print('Dims:')
print(dims)

print('-'*50)
print('Coords:')
print(coords)

print('-'*50)
print('Field axes:')
print(field)
```

## Plotting the Signal

Visualization is one of the most important parts of EPR data analysis.

We will generate a simple plot of the signal.

```python
plt.figure()
plt.title('Nitroxide Q-Band EDFS')
sl.plot(data.real)
plt.show()
```
